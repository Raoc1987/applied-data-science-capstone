````python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
````

## Complete Task 3: Calculate and Mark Railway Distance

### Subtask:
Calculate the distance from a launch site to the identified railway coordinates and add a PolyLine and marker displaying this distance on the map.

This requires re-initializing the map with all previous elements (launch sites, success/fail markers, and MousePosition) to ensure a complete and continuous visualization.



````python
import folium
import pandas as pd
import requests
from folium.plugins import MarkerCluster, MousePosition
from folium.features import DivIcon
from math import sin, cos, sqrt, atan2, radians

# --- Re-define calculate_distance function for robustness ---
def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance

# --- Data Loading and Initial Processing (Re-included for robustness) ---
url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/spacex_launch_geo.csv'
response = requests.get(url)

with open('spacex_launch_geo.csv', 'wb') as f:
    f.write(response.content)

spacex_df = pd.read_csv('spacex_launch_geo.csv')

# Select relevant sub-columns: `Launch Site`, `Lat(Latitude)`, `Long(Longitude)`, `class`
spacex_df = spacex_df[['Launch Site', 'Lat', 'Long', 'class']]
launch_sites_df = spacex_df.groupby(['Launch Site'], as_index=False).first()
launch_sites_df = launch_sites_df[['Launch Site', 'Lat', 'Long']]

# --- Prepare spacex_df with marker_color for Task 2 clustered markers ---
def assign_marker_color(launch_outcome):
    if launch_outcome == 1:
        return 'green'
    else:
        return 'red'
spacex_df['marker_color'] = spacex_df['class'].apply(assign_marker_color)

# Prepare launch_sites dictionary from launch_sites_df for initial markers (Task 1)
launch_sites = {
    row['Launch Site']: [row['Lat'], row['Long']]
    for index, row in launch_sites_df.iterrows()
}

# Calculate average coordinates for initial map centering
avg_lat = spacex_df['Lat'].mean()
avg_lon = spacex_df['Long'].mean()

# Initialize the map object once (using average coordinates for a good overview)
site_map = folium.Map(location=[avg_lat, avg_lon], zoom_start=5)

# Add blue circles and text labels for each unique launch site (Task 1)
for site_name, coords in launch_sites.items():
    folium.Circle(
        location=coords,
        radius=500, # meters
        color='blue',
        fill=True,
        fill_color='cyan',
        fill_opacity=0.5,
        popup=site_name
    ).add_to(site_map)
    folium.map.Marker(
        coords,
        icon=DivIcon(
            icon_size=(20,20),
            icon_anchor=(0,0),
            html=f'<div style="font-size: 12; color:#000000;"><b>{site_name}</b></div>',
        )
    ).add_to(site_map)

# Add MarkerCluster for success/failure launches (Task 2)
marker_cluster = MarkerCluster()
site_map.add_child(marker_cluster)

# Add individual launch markers to the cluster with color coding
for index, record in spacex_df.iterrows():
    marker = folium.Marker(
        location=[record['Lat'], record['Long']],
        popup=f"Launch Site: {record['Launch Site']}<br>Class: {record['class']}",
        icon=folium.Icon(color=record['marker_color'], icon='info-sign')
    )
    marker_cluster.add_child(marker)

# Add MousePosition plugin (Task 3)
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright',
    separator=' Long: ',
    empty_string='NaN',
    lng_first=False,
    num_digits=20,
    prefix='Lat:',
    lat_formatter=formatter,
    lng_formatter=formatter,
)
site_map.add_child(mouse_position)

# --- Subtask: Calculate distance to railway and mark it ---
# 1. Define the launch site coordinates for 'CCAFS LC-40'
launch_site_coords = [28.562302, -80.577356] # CCAFS LC-40

# 2. Assume RAILWAY_LAT and RAILWAY_LON (using placeholder for now)
# These coordinates are a hypothetical railway point near CCAFS LC-40, for demonstration
RAILWAY_LAT = 28.57216
RAILWAY_LON = -80.60627
railway_coords = [RAILWAY_LAT, RAILWAY_LON]

# 3. Calculate the distance
distance_railway = calculate_distance(launch_site_coords[0], launch_site_coords[1], railway_coords[0], railway_coords[1])

# 4. Create a folium.Marker at the railway_coords with DivIcon for distance
folium.Marker(
    location=railway_coords,
    icon=DivIcon(
        icon_size=(100, 30),
        icon_anchor=(0, 0),
        html=f'<div style="font-size: 12px; color:#d35400;"><b>{distance_railway:.2f} km (Railway)</b></div>'
    )
).add_to(site_map)

# Optional: Add a simple marker for the railway point itself for better visibility
folium.Marker(
    location=railway_coords,
    popup='Closest Railway',
    icon=folium.Icon(color='purple', icon='train', prefix='fa')
).add_to(site_map)

# 5. Create a folium.PolyLine object connecting the launch site and railway coordinates
folium.PolyLine(
    locations=[launch_site_coords, railway_coords],
    color='orange',
    weight=2.5,
    opacity=0.7
).add_to(site_map)

# 6. Display the updated site_map
site_map
````
